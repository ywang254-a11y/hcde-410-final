# A4: Final project preliminary proposal
*A Human-Centered Dara Study of Community Reporting in Seattle.*

## Motivation & problem
Many cities use digital reporting systems to collect information about problems in their communities. In Seattle, residents can use the Find It, Fix It app or other reporting channels to report issues such as graffiti, illegal dumping, and abandoned vehicles. These systems help the city understand where public services may be needed. However, I am interested in whether the information collected through these systems represents community needs equally across different neighborhoods.

For example, if one neighborhood has significantly more reports than another, does that necessarily mean it has more problems? Or could differences in reporting reflect other factors, such as residents' awareness of the system, access to technology, or willingness to report issues?

For this project, I want to explore how community reporting patterns vary across Seattle neighborhoods, with a particular focus on the different channels residents use to report public issues. I am also interetsted in whether these patterns are associated with socioeconomic characteristics, such as neighborhood income. From a human-centered persoective, I think this is important because the ways people interact with reporting systems may influence which community needs become visible in public data.

### My Primary Research Question

How do reporting channels vary across Seattle neighborhoods, and what might these differences reveal about participation in civic reporting systems?

To explore this question, I will focus on two research questions:

- **RQ1 -- Geographic & Channel Patterns:** How do reporting rates and reporting channels  vary across Seattle census tracts?

- **RQ2 -- Socioeconomic Associations:** How are reporting channel patterns associated with neighborhood socioeconomic characteristics, such as median household income?

My goal is to understand how different reporting channels may shape the community needs reflected in public data, while avoiding assumptions about individual residents based on neighborhood-level information.

## Data Used
The primary dataset I plan to use is the [City of Seattle's Customer Service Request dataset](https://cos-data.seattle.gov/City-Administration/Customer-Service-Requests/5ngg-rpne/about_data). It contains records of public service requests submitted through the city's reporting systems, including information about the request type, date, reporting method, status, and geographic location. The dataset is publicly available and listed under a Public Domain license. I will focus particularly on reporting methods to compare how different channels, such as mobile apps and phone calls, are used across Seattle neighborhoods.

I also plan to use [American Community Survey (ACS) data](https://catalog.data.gov/dataset/census-tract-top-50-american-community-survey-data) to examine neighborhood characteristics, such as population and medium household income. By combining these datasets with 2020 census tract boudnaries, I hope to compare reporting rates and reporting channel patterns across different areas and explore their relationships with socioeconomic characteristics. I will confirm the ACS data's reuse terms before including it in my public GitHub repository.

One important ethical consideration is that the reporting data only includes issues that people have reported. Fewer reports do not necessarily mean fewer problems, as differences may reflect barriers to participation or other factors not captured in the data. I will be careful not to make assumptions about individual residents based on neighborhood-level data.

## Unknowns & Dependencies 
One challenge will be combining the two datasets, especially matching service request locations with census tract boundaries using GeoPandas and checking for missing geographic information. I also need to determine which socioeconomic variables are available and whether the dataset clearly distinguishes between reporting channels, such as apps and phone calls. These factors may affect the scope of my analysis and whether I can complete it by the end of the quarter.

Another uncertainty is how to interpret differences in reporting rates without knowing the actual number of problems in each neighborhood. I plan to focus on identifying patterns and associations rather than making causal claims.
