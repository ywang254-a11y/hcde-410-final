*A Human-Centered Dara Study of Community Reporting in Seattle.*

# A4: Final project preliminary proposal

## Motivation & problem
Many cities use digital reporting systems to collect information about problems in their communities. In Seattle, residents can use the Find It, Fix It app or other reporting channels to report issues such as graffiti, illegal dumping, and abandoned vehicles. These systems help the city understand where public services may be needed. However, I am interested in whether the information collected through these systems represents community needs equally across different neighborhoods.

For example, if one neighborhood has significantly more reports than another, does that necessarily mean it has more problems? Or could differences in reporting reflect other factors, such as residents' awareness of the system, access to technology, or willingness to report issues?

For this project, I want to explore how community reporting patterns vary across Seattle neighborhoods, with a particular focus on the different channels residents use to report public issues. I am also interetsted in whether these patterns are associated with socioeconomic characteristics, such as neighborhood income. From a human-centered persoective, I think this is important because the ways people interact with reporting systems may influence which community needs become visible in public data.

### My Primary Research Question

How do reporting channels vary across Seattle neighborhoods, and what might these differences reveal about participation in civic reporting systems?

To explore this question, I will focus on two research questions:

- **RQ1 -- Geographic & Channel Patterns:** How do reporting rates and reporting channels  vary across Seattle census tracts?

- **RQ2 -- Socioeconomic Associations:** How are reporting channel patterns associated with neighborhood socioeconomic characteristics, such as median household income?

My goal is to understand how different reporting channels may shape the community needs reflected in public data, while avoiding assumptions about individual residents based on neighborhood-level information.

## Data Used
The primary dataset I plan to use is the [City of Seattle's Customer Service Request dataset](https://cos-data.seattle.gov/City-Administration/Customer-Service-Requests/5ngg-rpne/about_data). It contains records of public service requests submitted through the city's reporting systems, including information about the request type, date, reporting method, status, and geographic location. The dataset is publicly available and listed under a Public Domain license. I will focus particularly on reporting methods to compare how different channels, such as mobile apps and phone calls, are used across Seattle neighborhoods.

I also plan to use [American Community Survey (ACS) data](https://catalog.data.gov/dataset/census-tract-top-50-american-community-survey-data) to examine neighborhood characteristics, such as population and medium household income. By combining these datasets with 2020 census tract boudnaries, I hope to compare reporting rates and reporting channel patterns across different areas and explore their relationships with socioeconomic characteristics. I will confirm the ACS data's reuse terms before including it in my public GitHub repository.

One important ethical consideration is that the reporting data only includes issues that people have reported. Fewer reports do not necessarily mean fewer problems, as differences may reflect barriers to participation or other factors not captured in the data. I will be careful not to make assumptions about individual residents based on neighborhood-level data.

## Background / related work 
I reviewed several research that examined how civic reporting systems, such as 311 services, support public participation and how reporting patterns may differ acorss communities.

The first study by Clark, Brudney, and Jang (2013) focused on Boston's 311 system to understand whether residents from different socioeconomic backgrounds participated equally in reporting public service issues. Their findings showed that neighborhoods with lower income generally had lower levels of 311 participation, although smartphone reporting showed a different pattern. This made me curious about whether different reporting channels show similar patterns across Seattle neighborhoods.

Another study by Lee et al. (2021) also examined Boston's 311 system, which examined how differences in technology use and reporting behavior may contribute to bias in civic data. The researchers highlighted that data collected throught these systems may not fully represent community needs, especially when some groups contribute more data than others. This connects to my project because the way people interact with reporting systems may influence which community issues become visible in the data.

These studies help shape my research in two ways. First, they suggest that I should examine different reporting channels separately instead of treating all service requests as the same. Also, they show the improtance of considering neighborhood socioeconimic characteristics when interpreting reporting patterns. Building on these studies, I want to further explore whether similar patterns appear in Seattle, while recognizing that reporting data alone cannot explain individual residents' experiences or motivations.

## Methodology

For this project, I plan to use Python in Jupyter Notebook to conduct descriptive statistical analysis, geographic analysis, and correlation analysis. I will use pandas and GeoPandas to prepare the data and explore reporting patterns across Seattle census tracts.

### 1. Data Preparation

I will focus on Seattle service requests created between January and December 2025. First, I will use pandas to clean the dataset, check for missing values and duplicate records, and organize requests by reporting channel and request type. After that, I will use GeoPandas to match service request coordiniates with Seattle's 2020 census tract boundaries through a spatial join. This will allow me to group reports by geographic area. I also plan to use 2020-2024 ACS 5 year estimates, including total population and median household income. Before combining the datasets, I will check whether the census tract identifiers match and whether the selected ACS records contain valid values. 

### 2. Preliminary Feasibility Testing

Before conducting the analysis, I created a separate notebook, [Data_Test.ipynb](Data_Test.ipynb), to check whether the datasets and methods could support my research.

The tests confirmed that the service request dataset distinguishes between Find It Fix It App and Phone reports and includes different request types. I also checked the availability of geographic coordinates for each reporting channel. To test the geographic analysis, I performed a spatial join using 300 sample requests, including 150 App reports and 150 Phone reports. Of these, 296 records (98.67%) were successfully matched to census tracts. Also, I confirmned that the ACS dataset includes population and median household income fields. 

The testing results suggest that my proposed methods are feasible. The full data processing and statistical analysis will be completed in the next stage of the project.

### 3. RQ1 -- Geographic & Channel Patterns

To answer RQ1, I will calculate the number of reports submitted through different channels in each census tract. I will also use ACS population data to calculate reporting rates per 1000 residents, since areas have different population sizes.

For the App and Phone comparison, I will calculate each channel's share of the reportrs submitted through these two channels. I also plan to compare selected request types, such as abandoned vehicles and graffiti, since different types of issues may have different reporting patterns.

I chose descriptive statistics and geographic analysis because they can help identify differences across areas and make reporting patterns easier to understand. I will use bar charts and choropleth maps to present these results.

### 4. RQ2 -- Socioeconomic Associations

To answer RQ2, I will examine whether reporting channel patterns are associated with census tract median household income.

I plan to use scatter plots and Spearman correlation to explore relationships between income and reporting channel proportions or population-normalized reporting rates. I chose Spearman correlation because it can measure monotonic relationships without assuming that the relationship is linear.

I will compare the results across App and Phone reports to see whether they show different associations with neighborhood income. These comparisons will focus on assiciations and will not establish causal relationships.

### 5. Presentation & Limitations

I plan to present my findings using maps, charts, and short written explanations. I will organize the results around my two research questions and explain what each analysis shows.

One limitation is that service request data only includes reported issues and does not capture people who have never used the system. Also, the location of a report does not necessarily represent where the person submitting it lives. Differences in geographic data completeness across reporting channels may also affect the analysis. Population-based reporting rates may also affected by visitors and commuters, especially in areas with relatively few residents.

I will document my data cleaning and analysis steps in Jupyter Notebook and share the code and links to the data sources through Github to support reproducibility.
