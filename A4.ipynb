# A4: Final project preliminary proposal
*A Human-Centered Dara Study of Community Reporting in Seattle.*

## Motivation & problem
Many cities use digital reporting systems to collect information about problems in their communities. In Seattle, residents can use the Find It, Fix It app or other reporting channels to report issues such as graffiti, illegal dumping, and abandoned vehicles. These systems help the city understand where public services may be needed. However, I am interested in whether the information collected through these systems represents community needs equally across different neighborhoods.

For example, if one neighborhood has significantly more reports than another, does that necessarily mean it has more problems? Or could differences in reporting reflect other factors, such as residents' awareness of the system, access to technology, or willingness to report issues?

For this project, I want to explore how community reporting patterns vary across Seattle neighborhoods and whether these differences are associated with socioeconomic characterstics, particularly income. From a human-centered perspective, I think this is important because data collected through public reporting systems may influence how we understand community needs, even though not everyone's experiences are necessarily captured.

### My Primary Research Question

How do community reporting patterns vary across Seattle neighborhoods, and what might these differences reveal about the visibility of community needs in civic data?

To explore this question, I am considering three more specific research questions:

- **RQ1 -- Geographic Differences:** How do service request volumes and types vary across Seattle neighborhoods?
- **RQ2 -- Socioeconomic Patterns:** How are neighborhood reporting rates associated with socioeconomic characteristics such as income?
- **RQ3 -- Reporting Channels:** How do reporting methods vary across neighborhoods, and are these patterns associated with neighborhood characteristics?

My goal is to understand what these reporting patterns reveal and what they might overlook, while avoiding assumptions about individual residents based on neighborhood-level data.

## Data Used
The primary dataset I plan to use is the [City of Seattle's Customer Service Request dataset](https://cos-data.seattle.gov/City-Administration/Customer-Service-Requests/5ngg-rpne/about_data). It contains records of public service requests submitted through the city's reporting systems, including information about the request type, date, reporting method, status, and geographic location. The dataset is publicly available and listed under a Public Domain license.

I also plan to [use American Community Survey(ACS) data](https://catalog.data.gov/dataset/census-tract-top-50-american-community-survey-data) to examine neighborhood characteristics, such as population and potentially household income. This dataset includes census tract-level demographic information. By combining these datasets, I hope to compare reporting rates across neighborhoods and explore their relationships with socioeconomic characteristics. The ACS dataset is publicly available through the City of Seattle's open data portal, and I will confirm its reuse terms before including it in my public GitHub repository.

One important ethical consideration is that the reporting data only includes issues that people have reported. Fewer reports do not necessarily mean fewer problems, as differences may reflect barriers to participation or other factors not captured in the data. I will be careful not to make assumptions about individual residents based on neighborhood-level data.

## Unknowns & Dependencies 
One challenge will be combining the two datasets, especially matching service request locations with census tract boundaries and checking for missing geographic information. I also need to determine which socioeconomic variables are available and appropriate for comparison. These factors may affect the scope of my analysis and whether I can complete it by the end of the quarter.

Another uncertainty is how to interpret differences in reporting rates without knowing the actual number of problems in each neighborhood. I plan to focus on identifying patterns and associations rather than making causal claims.
